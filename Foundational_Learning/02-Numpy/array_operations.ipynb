Deep and Shallow Copy

```python
a=np.arange(1,21)
a
```

```python
slice=a[:5]
slice = slice * 10
slice
```

```python
a # in lists whenever we slicing any portion of it.. it effects the original list where as in Arrays we are slicing any portion of array it creates different location/space in RAM so original array doesn't effected by slicing.This is the Concept of Shallow Copy
```

```python
a[4:6] * 10
```

```python
a
```

```python
a
b=a # here total array 'a' is copied by 'b' 
```

```python
b[0]=100
b
```

```python
a # so both 'a' and 'b' have same location that's why changes in array 'b' effects array 'a'
```

```python
a1=np.arange(1,21)
a1
```

```python
a1
b1=a1.copy() # this enables Shallow Copy so the changes in array 'b1' doesn't effect on array 'a1'
```

```python
b1[1]=200
b1
```

```python
a1
```

Stacking Array

```python
a=np.array([1,2,3,4])
b=np.array([5,6,7,8])
c=np.array([9,10,11,12])
```

```python
np.vstack((a,b,c)) # vertical stack
```

```python
np.hstack((a,b,c)) # horizontal stack
```

```python
np.column_stack((a,b,c)) # column stack
```
